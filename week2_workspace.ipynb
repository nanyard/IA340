# Diamond Price Analysis

```python
import pandas as pd

file_path = '/content/drive/MyDrive/IA340/diamonds.csv'
df = pd.read_csv(file_path)

# Display the first five rows
print(df.head().to_markdown(index=False))
```

This cell loads the `diamonds.csv` dataset into a pandas DataFrame and displays the first 5 rows to get a quick overview of the data structure.

```python
average_price_per_color = df.groupby('COLOR')['PRICE'].mean().reset_index()
print(average_price_per_color.to_markdown(index=False))
```

This cell calculates the average price of diamonds for each color category. It groups the DataFrame by the 'COLOR' column and then computes the mean of the 'PRICE' for each group.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

These libraries are imported for data visualization. `matplotlib.pyplot` is a plotting library and `seaborn` is built on top of matplotlib to provide a high-level interface for drawing attractive statistical graphics.

```python
# Create a scatter plot to show price, weight, and clarity
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='WEIGHT', y='PRICE', hue='CLARITY', palette='viridis', s=100, alpha=0.7)

# Add titles and labels
plt.title('Diamond Price vs. Weight by Clarity')
plt.xlabel('Weight (Carats)')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Clarity', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

This cell generates a scatter plot to visualize the relationship between diamond 'WEIGHT' (size), 'PRICE', and 'CLARITY'. Each point represents a diamond, with its color indicating its clarity level. This helps to observe how price changes with weight for different clarity grades.
